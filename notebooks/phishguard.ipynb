# PhishGuard walkthrough

Lexical URL features, then logistic regression from scratch vs sklearn. Train with `python -m phishguard train` first if `reports/metrics.json` is missing.

```python
from phishguard.features import extract

for url in [
    "https://www.wikipedia.org",
    "http://192.168.1.20/paypal/login/verify-account",
]:
    print(url)
    print(extract(url))
    print()
```

Held-out test numbers after threshold tuning on the validation set.

```python
import json
from pathlib import Path

m = json.loads(Path("reports/metrics.json").read_text())
for name in ("numpy", "sklearn"):
    s = m[name]
    print(f"{name:8}  thr={s['threshold']:.2f}  P={s['precision']:.3f}  "
          f"R={s['recall']:.3f}  F1={s['f1']:.3f}  AUC={s['roc_auc']:.3f}")
```
